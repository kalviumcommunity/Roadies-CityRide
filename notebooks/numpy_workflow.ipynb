# NumPy Vectorised Computation Workflow

Demonstrating vectorised numerical operations for performance-critical calculations.

```python
import numpy as np
import pandas as pd

from roadies.analysis.numpy_workflow import (
    vectorised_demand_supply_ratio,
    baseline_demand_supply_ratio,
    vectorised_percentage_change,
    vectorised_risk_classification,
    vectorised_zscore,
    vectorised_normalise,
    benchmark_operations,
)
```

```python
# Create sample data
np.random.seed(42)
n = 10000
demand = np.random.uniform(100, 1000, n)
supply = np.random.uniform(50, 500, n)
print(f'Dataset: {n} rows')
```

```python
# Vectorised vs baseline comparison
vec_result = vectorised_demand_supply_ratio(demand, supply)
base_result = baseline_demand_supply_ratio(pd.Series(demand), pd.Series(supply)).values

print(f'Vectorised: {vec_result[:5]}')
print(f'Baseline:   {base_result[:5]}')
print(f'Max diff:   {np.nanmax(np.abs(vec_result - base_result))}')
```

```python
# Risk classification
surge = np.random.uniform(1, 3, n)
risk = vectorised_risk_classification(surge, 1.5, 2.0, 2.5)
print(f'Risk distribution: {np.bincount(risk)}')
```

```python
# Benchmark
results = benchmark_operations(n_rows=100000, n_iterations=5)
for r in results:
    print(f'{r.operation}: {r.speedup:.2f}x speedup')
```
